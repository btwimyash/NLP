```python
# Import required libraries
import nltk
from nltk.tokenize import WhitespaceTokenizer, wordpunct_tokenize, TreebankWordTokenizer, TweetTokenizer
from nltk.stem import PorterStemmer, SnowballStemmer
from nltk.stem import WordNetLemmatizer
```

```python
# Download required NLTK resources 
# nltk.download('punkt')
# nltk.download('wordnet')
# nltk.download('omw-1.4')

# Sample text
text = "I love programming in Python! It's awesome. #coding @python_dev http://python.org"

print("Original Text:")
print(text)
print("\n---------------- TOKENIZATION ----------------")
```

```python
# 1. Whitespace Tokenization
wt = WhitespaceTokenizer()
print("Whitespace Tokenizer:")
print(wt.tokenize(text))

# 2. Punctuation-based Tokenization
print("\nPunctuation-based Tokenizer:")
print(wordpunct_tokenize(text))
```

```python
# 3. Treebank Tokenization
tbt = TreebankWordTokenizer()
print("\nTreebank Tokenizer:")
print(tbt.tokenize(text))

# 4. Tweet Tokenization
tweet_tokenizer = TweetTokenizer()
print("\nTweet Tokenizer:")
print(tweet_tokenizer.tokenize(text))
```

```python
# 5. Multi-Word Expression (MWE) Tokenization
from nltk.tokenize import MWETokenizer
mwe = MWETokenizer([('New', 'York'), ('machine', 'learning')])
sample_mwe_text = "I live in New York and study machine learning"
print("\nMWE Tokenizer:")
print(mwe.tokenize(sample_mwe_text.split()))

print("\n---------------- STEMMING ----------------")

# Porter Stemmer
porter = PorterStemmer()
print("Porter Stemmer:")
print([porter.stem(word) for word in ["running", "studies", "better"]])

# Snowball Stemmer
snowball = SnowballStemmer("english")
print("\nSnowball Stemmer:")
print([snowball.stem(word) for word in ["running", "studies", "better"]])
```

```python
print("\n---------------- LEMMATIZATION ----------------")

# Lemmatization
lemmatizer = WordNetLemmatizer()
print("Lemmatization:")
print([lemmatizer.lemmatize(word, pos='v') for word in ["running", "studies", "better"]])
```
