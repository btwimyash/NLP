```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.preprocessing import normalize
from gensim.models import Word2Vec
```

```python
# Sample data
documents = [
    "I love natural language processing",
    "Natural language processing is interesting",
    "I love learning NLP"
]
```

```python
# ---------------- BAG OF WORDS ----------------

# Count Occurrence
count_vectorizer = CountVectorizer()
bow_count = count_vectorizer.fit_transform(documents)

print("Bag of Words - Count Occurrence:")
print(count_vectorizer.get_feature_names_out())
print(bow_count.toarray())
```

```python
# Normalized Count Occurrence
bow_normalized = normalize(bow_count, norm='l2')

print("\nBag of Words - Normalized Count Occurrence:")
print(bow_normalized.toarray())
```

```python
# ---------------- TF-IDF ----------------

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

print("\nTF-IDF Feature Names:")
print(tfidf_vectorizer.get_feature_names_out())
print(tfidf_matrix.toarray())
```

```python

# ---------------- WORD2VEC ----------------

tokenized_docs = [doc.split() for doc in documents]

word2vec_model = Word2Vec(
    sentences=tokenized_docs,
    vector_size=50,
    window=3,
    min_count=1
)

print("\nWord2Vec Embedding for word 'language':")
print(word2vec_model.wv["language"])
```
